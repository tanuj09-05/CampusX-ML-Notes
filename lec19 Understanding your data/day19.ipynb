# 4. Are there any missing values?
"""
Missing values ka matlab: kuch cells empty hain ya NaN hain
"""

```python
df.isnull().sum()
"""
Is line ko parts me samjho:
df.isnull() har cell ke liye check karta hai ki value missing hai ya nahi.

Agar missing hai to True, warna False.

.sum() har column ke True values count karta hai. Python me True = 1 aur False = 0 treat hota hai.

Output:
PassengerId      0
Survived         0
Pclass           0
Name             0
Sex              0
Age            177
SibSp            0
Parch            0
Ticket           0
Fare             0
Cabin          687
Embarked         2


Meaning:
Age me 177 missing values hain.
Cabin me 687 missing values hain, bahut zyada missing hai.
Embarked me 2 missing values hain.

Baaki columns complete hain.
"""
```

# 5. How does the data look mathematically?
"""
numerical columns ka mathematical summary kaisa hai?
Yani average, minimum, maximum, median etc.
"""

```python
df.describe()
"""
df.describe() numerical columns ka statistical summary deta hai.

Output me ye rows hoti hain:
count: non-missing values kitni hain.
mean: average value.
std: standard deviation, yani data kitna spread hai.
min: minimum value.
25%: first quartile.
50%: median.
75%: third quartile.
max: maximum value.

Example:
Survived ka mean 0.383838 hai.
Matlab around 38.38% passengers survived.
Age ka mean around 29.7 hai.
Average age approx 30 years hai.
Fare ka max 512.3292 hai.

Matlab sabse expensive ticket ka fare 512.3292 tha.
"""
```
